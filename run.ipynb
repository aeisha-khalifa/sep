```python
import os, shutil, zipfile

# Mount drive
from google.colab import drive
drive.mount('/content/drive')

# Extract zip with correct paths
with zipfile.ZipFile('/content/drive/MyDrive/sep_colab.zip', 'r') as z:
    for member in z.namelist():
        target = member.replace('\\', '/')
        target_path = os.path.join('/content', target)
        if not os.path.basename(target):
            os.makedirs(target_path, exist_ok=True)
        else:
            os.makedirs(os.path.dirname(target_path), exist_ok=True)
            with z.open(member) as src, open(target_path, 'wb') as dst:
                dst.write(src.read())

# Fix any remaining backslash names
for name in os.listdir('/content'):
    if '\\' in name:
        src = os.path.join('/content', name)
        dst = os.path.join('/content', name.replace('\\', '/'))
        os.makedirs(os.path.dirname(dst), exist_ok=True)
        shutil.move(src, dst)

print("main.py exists:", os.path.exists('/content/main.py'))
print("exp exists:", os.path.exists('/content/exp'))
```
